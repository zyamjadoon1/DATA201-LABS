```python
#9(a)
import pandas as pd
Auto = pd.read_csv("Auto.csv")
print(Auto.head())
```

```python
# Quantitative: mpg, horsepower, weight, acceleration, displacement.
# Qualitative: cylinders, year, origin, name
```

```python
#(b)
# Re-load with '?' treated as missing (NaN)
Auto = pd.read_csv("Auto.csv", na_values='?')

# Drop rows with any missing values
Auto = Auto.dropna()

print(Auto.shape)

quant_vars = ['mpg', 'displacement', 'horsepower', 'weight', 'acceleration']
for col in quant_vars:
  print(col, ":", Auto[col].min(), "-", Auto[col].max())
```

```python
#(c)
Auto[quant_vars].describe()
```

```python
#(d)
Auto_subset = Auto.drop(Auto.index[10:85])

for col in quant_vars:
    print(col, ":",
          "range =", Auto_subset[col].min(), "-", Auto_subset[col].max(),
          " mean =", Auto_subset[col].mean(),
          " std =", Auto_subset[col].std())
```

```python
#(e)
import matplotlib.pyplot as plt
import pandas as pd

pd.plotting.scatter_matrix(Auto[['mpg', 'displacement', 'horsepower', 'weight', 'acceleration']], figsize=(10, 10))
plt.show()
```

```python
# 9(f)
# Based on the scatterplots, weight, displacement, and horsepower all show
# a clear negative relationship with mpg, suggesting they would be useful
# predictors. Cylinders also shows a pattern (fewer cylinders → higher mpg).
# Acceleration shows a weaker, less clear relationship with mpg.
```
